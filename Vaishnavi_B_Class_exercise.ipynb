```python
SYSTEM_PROMPT="You are an Sentiment Analyzer Assistant, your task is to classify the given reviews, also extract the structured output according to the attached schema"
PROMPT="Analyze the sentiment of each review, categorize it as positive, neutral, or negative, and store the result. Extract key pieces of information from each review, such as the pros and cons mentioned, and the specific features the reviewer liked or disliked (e.g., vehicle performance, comfort, price)."
```

```python
from pydantic import BaseModel, Field

class ReviewSchema(BaseModel):
  Review_Date: str =Field(description="The date of review"),
  Vehicle_Title: str =Field(description="The full name/model of the car"),
  Review_Text: str =Field(description="The compelete review provided by the user"),
  Sentiment : str =Field(description="The overall sentiment : Positive, Negative or Neutral"),
  Pros: str =Field(description="All pros separated by commas, and each of them in quotes, return None if not present "),
  Cons: str =Field(description="All cons separated by commas, and each of them in quotes, return None if not present "),
  Liked_Features: list =Field(description="All features liked by the reviewer, empty if not present "),
  Disliked_Features: list =Field(description="All features liked by the reviewer, empty if not present ")

```
